## Data Cleaning

**7. Remove rows where `YIELD_KG` is missing**

```python
crop_df = crop_df.dropna(subset=['YIELD_KG'])
crop_df.shape
```

**8. Fill missing `AREA_ACRES` with the mean**

```python
crop_df['AREA_ACRES'] = crop_df['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())
```

**9. Replace missing `DISTRICT` with 'Unknown'**

```python
crop_df['DISTRICT'] = crop_df['DISTRICT'].fillna('Unknown')
```

**10. Replace missing `FERTILIZER_USED` with 'No'**

```python
crop_df['FERTILIZER_USED'] = crop_df['FERTILIZER_USED'].fillna('No')
```

**11. Replace missing `YIELD_VALUE` with the mean**

```python
crop_df['YIELD_VALUE'] = crop_df['YIELD_VALUE'].fillna(crop_df['YIELD_VALUE'].mean())
```

**12. Convert `HARVEST_DATE` to datetime**

```python
crop_df['HARVEST_DATE'] = pd.to_datetime(crop_df['HARVEST_DATE'], format='%d/%m/%Y', errors='coerce')
```

**13. Remove the letter 'F' from `FARM_ID` and convert to integer**

```python
crop_df['FARM_ID'] = crop_df['FARM_ID'].str.replace('F', '').astype(int)
crop_df.head()
```

**14. Re-check missing values after cleaning**

```python
crop_df.isnull().sum()
```

**15. Re-check data types after cleaning**

```python
crop_df.dtypes
```

## Data Analysis and Visualization

**16. Total yield of each crop type**

```python
crop_yield = crop_df.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
crop_yield
```

**17. Bar chart – Total Yield by Crop Type (KG)**

```python
crop_yield.plot(kind='bar', legend=False, figsize=(8, 5), color='green')
plt.title('Total Yield by Crop Type (KG)')
plt.xlabel('Crop Type')
plt.ylabel('Yield (KG)')
plt.xticks(rotation=0)
plt.show()
```

**18. Pivot table of total yield by harvest date, then the trend line chart**

```python
pivot = crop_df.pivot_table(values='YIELD_KG', index='HARVEST_DATE', aggfunc='sum')
pivot.head(15)
```

```python
pivot.plot(kind='line', legend=False, figsize=(11, 5))
plt.title('Yield Trend Over Time (KG)')
plt.xlabel('Harvest Date')
plt.ylabel('Yield (KG)')
plt.tight_layout()
plt.savefig('crop_yield_trend.png')
plt.show()
```
